## 3.2 Windowing + FFT Bandpower Features

This section creates window-level features from raw EEG channels (TP9, AF7, AF8, TP10) using the **P2 baseline workflow**:

- **Window first, split later**: we generate windows within each Participant×Class segment, then do a stratified train/test split on the resulting window-level samples.
- **Cleaning before extraction**: we coerce EEG columns to numeric and drop invalid rows before feature extraction.
- **Sliding windows**: window length `win=64` samples, step `step=16` samples (overlap).
- **FFT-only features**: compute per-window **bandpower** using FFT (no Welch/PSD, no time-domain stats).
- **Output features**: 4 channels × 5 bands = **20 FFT bandpower features** per window (log-transformed).

**Note (for P2 panel):** this baseline split is *not participant-disjoint*. Participant-disjoint validation (e.g., leave-one-participant-out) and other advanced formulations are planned for the final defense stage.

### 3.2.1 Windowed EEG Feature Extraction

````python
eeg_cols = ['RAW_TP9','RAW_AF7','RAW_AF8','RAW_TP10']

fs_hz = 256  # Sampling rate (Hz) used for FFT frequency axis

window_samples = 64   # Multiple windows per segment (~154 rows each)
step_samples = 16  # overlap: step=16 per P2 baseline (was 32)
win = int(window_samples)
step = int(step_samples)

bands = {
    'delta': (1, 4),
    'theta': (4, 8),
    'alpha': (8, 13),
    'beta' : (13, 30),
    'gamma': (30, 45),
}

# Column names for classical ML features (4 channels × 5 bands)
feature_cols = [f"{ch}_{b}" for ch in eeg_cols for b in bands.keys()]


def bandpower_psd_welch(x, fs, fmin, fmax, nperseg=None):
    """
    Computes the absolute bandpower in a specific frequency range
    using Welch's Power Spectral Density (PSD) method.
    """
    # If segment length isn't provided, default to window length or 32 samples
    if nperseg is None:
        nperseg = min(len(x), 32)

    # Compute the Power Spectral Density (PSD)
    freqs, psd = welch(x, fs=fs, window='hann', nperseg=nperseg, noverlap=nperseg//2)

    # Find the indices corresponding to our target frequency band
    idx_band = np.logical_and(freqs >= fmin, freqs <= fmax)

    # Frequency resolution (df)
    df = freqs[1] - freqs[0]

    # Integrate the area under the PSD curve for the selected frequency band
    _trapz = getattr(np, 'trapezoid', np.trapz)
    band_power = _trapz(psd[idx_band], dx=df)

    return band_power

required_cols = set(['Participant_id', 'Classification_id', *eeg_cols])
missing = required_cols.difference(csvfile.columns)


def clean_for_extraction(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()

    # 1) Basic cleaning
    out = out.replace('#NUM!', np.nan)
    for c in eeg_cols:
        out[c] = pd.to_numeric(out[c], errors='coerce')

    out = out.dropna(subset=eeg_cols + ['Participant_id', 'Classification_id']).reset_index(drop=True)
    out['Participant_id'] = out['Participant_id'].astype(int)
    out['Classification_id'] = out['Classification_id'].astype(int)

    # 2) ASR Artifact Removal
    raw_data = out[eeg_cols].to_numpy(dtype=np.float64)  # (n_samples, n_channels)

    ASR_cls = ASR

    def _make_asr():
        # Try a few constructor signatures across meegkit versions.
        try:
            return ASR_cls(sfreq=fs_hz, cutoff=20, win_len=0.5)
        except TypeError:
            pass
        try:
            return ASR_cls(sfreq=fs_hz, cutoff=20)
        except TypeError:
            pass
        try:
            return ASR_cls(fs_hz, cutoff=20, win_len=0.5)
        except TypeError:
            pass
        return ASR_cls(fs_hz)

    err1 = None

    # Attempt 1: (n_channels, n_samples)
    data_for_asr = raw_data.T
    try:
        asr = _make_asr()
        asr.fit(data_for_asr)
        cleaned_for_asr = asr.transform(data_for_asr)
        cleaned = cleaned_for_asr.T
    except Exception as e:
        err1 = e
        cleaned = None

    # Attempt 2: (n_samples, n_channels)
    if cleaned is None:
        try:
            asr = _make_asr()
            asr.fit(raw_data)
            cleaned = asr.transform(raw_data)
        except Exception as e2:
            raise RuntimeError(
                "ASR failed with both orientations. "
                f"Error(ch×samples)={err1}; Error(samples×ch)={e2}"
            )

    out.loc[:, eeg_cols] = cleaned.astype(np.float32)
    print(f"ASR applied: input={raw_data.shape} (samples, ch)")
    return out


clean_df = clean_for_extraction(csvfile)

# Convenience alias if later cells want the ASR-cleaned dataframe explicitly
csvfile_asr = clean_df

print('After cleaning for extraction:', clean_df.shape)
print('Using win=', win, 'step=', step)


def build_windowed_features(df: pd.DataFrame):
    feature_rows = []
    y = []
    groups = []
    seq_windows = []

    group_cols = ['Participant_id', 'Classification_id']

    seg_lengths = [len(g) for _, g in df.groupby(group_cols, sort=False)]
    if len(seg_lengths) > 0:
        print('Segment length summary (rows per Participant×Class):')
        print(pd.Series(seg_lengths).describe())

    for (pid, class_id), g in df.groupby(group_cols, sort=False):
        eeg = g[eeg_cols].to_numpy(dtype=np.float32)
        n = eeg.shape[0]
        if n < win:
            continue

        for start in range(0, n - win + 1, step):
            seg = eeg[start:start + win]  # (win, 4)
            if not np.isfinite(seg).all():
                continue

            # Sequence window (for sequence models)
            seq_windows.append(seg)

            # FFT bandpower features (for classical models)
            feats = []
            for ch in range(seg.shape[1]):
                xch = seg[:, ch]
                for (fmin, fmax) in bands.values():
                    bp = bandpower_psd_welch(xch, fs=fs_hz, fmin=fmin, fmax=fmax, nperseg=32)
                    feats.append(np.log10(bp + 1e-8))

            feature_rows.append(feats)
            y.append(int(class_id))
            groups.append(int(pid))

    x_band = np.asarray(feature_rows, dtype=np.float32)
    X_seq = np.asarray(seq_windows, dtype=np.float32)
    y = np.asarray(y, dtype=np.int32)
    groups = np.asarray(groups, dtype=np.int32)
    return x_band, X_seq, y, groups


x_band_all, X_seq_all, y_all, groups_all = build_windowed_features(clean_df)

print('All windows:', 'x_band:', x_band_all.shape, 'X_seq:', X_seq_all.shape, 'y:', y_all.shape)

# Split windows into train/test
TEST_SIZE = float(globals().get('TEST_SIZE', 0.30))
RANDOM_STATE = int(globals().get('RANDOM_STATE', 5))
SPLIT_STRATEGY = 'group'

if SPLIT_STRATEGY == 'group':
    # Participant-disjoint split (more strict) — optional for later experiments
    from sklearn.model_selection import GroupShuffleSplit

    gss = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=RANDOM_STATE)
    train_idx, test_idx = next(gss.split(x_band_all, y_all, groups=groups_all))

    x_band_train, x_band_test = x_band_all[train_idx], x_band_all[test_idx]
    X_train_seq, X_test_seq = X_seq_all[train_idx], X_seq_all[test_idx]
    y_train_arr, y_test_arr = y_all[train_idx], y_all[test_idx]
    groups_train, groups_test = groups_all[train_idx], groups_all[test_idx]

    print('Split strategy: group (participant-disjoint)')
else:
    x_band_train, x_band_test, X_train_seq, X_test_seq, y_train_arr, y_test_arr, groups_train, groups_test = train_test_split(
        x_band_all,
        X_seq_all,
        y_all,
        groups_all,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE,
        stratify=y_all,
    )
    print('Split strategy: random (stratified by class)')

print('Train shapes:', 'x_band_train:', x_band_train.shape, 'X_train_seq:', X_train_seq.shape, 'y_train:', y_train_arr.shape)
print('Test  shapes:', 'x_band_test :', x_band_test.shape, 'X_test_seq :', X_test_seq.shape, 'y_test :', y_test_arr.shape)
print('Feature columns used for FFT features:', feature_cols)
````

# **4. Scaling (Train-only)**

````python
# ==============================================================================
# LOSO Cross-Validation — Deep Learning Models (Sequence + EEG conv)
# ==============================================================================
# Runs every DL architecture through Leave-One-Subject-Out CV using the same
# x_band_all / X_seq_all / y_all / groups_all arrays built in the windowing cell.
#
# Two input formats are covered inside one loop:
#   - seq   : (n_windows, timesteps, n_channels)  — LSTM, RNN, CNN1D, Mambular, Transformer, BERT, CNN-BiLSTM
#   - eeg2d : (n_windows, n_channels, timesteps, 1) — EEGNet, DeepConvNet
#
# Each fold:
#   1. Splits windows by participant (no leakage).
#   2. Normalises sequences per-channel using TRAINING statistics only.
#   3. Rebuilds the model from scratch (fresh weights) to avoid fold-to-fold bleed.
#   4. Uses a held-out validation participant from the training set for early stopping.
#
# Bug fixes vs original version:
#   - clear_session() is called after each model with a unique model._name to avoid
#     layer-name collisions. Models are renamed with fold index to stay unique.
#     del model + gc.collect() + clear_session() handles memory on 16GB RAM.
#   - EarlyStopping moved INSIDE the per-model loop: the callback is stateful
#     (tracks best, wait, best_weights). Sharing one instance across 9 models per
#     fold caused stale state — models 2-9 could stop on epoch 1 or restore wrong weights.
#   - val_pid selection fixed: uses (fold+1) % n to avoid repeating p0 for both
#     fold 0 and fold 30.
# ==============================================================================

_timesteps   = int(X_seq_all.shape[1])
_n_channels  = int(X_seq_all.shape[2])
_num_classes = int(len(np.unique(y_all)))

print(f"LOSO DL setup  |  timesteps={_timesteps}  channels={_n_channels}  "
      f"classes={_num_classes}  windows={len(y_all)}  participants={len(np.unique(groups_all))}")

# ── Model factory functions ───────────────────────────────────────────────────
# Each takes (timesteps, n_channels, num_classes) and returns an uncompiled Model.
# Also defined once outside loops — they contain no state, only graph-building logic.

def _build_lstm(ts, nc, nc_):
    return Sequential([
        Input(shape=(ts, nc)),
        Bidirectional(LSTM(64, return_sequences=False)),
        Dropout(0.3),
        Dense(nc_, activation='softmax'),
    ])

def _build_cnn1d(ts, nc, nc_):
    return Sequential([
        Input(shape=(ts, nc)),
        Conv1D(64, 7, padding='same', kernel_regularizer=l2(1e-4)),
        BatchNormalization(), Activation('relu'), MaxPooling1D(2), Dropout(0.20),
        Conv1D(128, 5, padding='same', kernel_regularizer=l2(1e-4)),
        BatchNormalization(), Activation('relu'), MaxPooling1D(2), Dropout(0.30),
        Conv1D(256, 3, padding='same', kernel_regularizer=l2(1e-4)),
        BatchNormalization(), Activation('relu'), GlobalAveragePooling1D(), Dropout(0.40),
        Dense(128, activation='relu'), Dropout(0.30),
        Dense(nc_, activation='softmax'),
    ])

def _build_cnn_bilstm(ts, nc, nc_):
    # Improved architecture — matches Section 6.9 with stronger regularization
    # Note: Bidirectional(LSTM(64)) outputs 128 features (64 forward + 64 backward).
    # All residual projections must match this dim (128).
    k1, k2 = min(7, ts), min(5, ts)
    inp = Input(shape=(ts, nc))
    # CNN feature extraction (64 filters, with BN + L2)
    x = Conv1D(64, k1, padding='same', kernel_regularizer=l2(1e-4))(inp)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Dropout(0.2)(x)
    x = Conv1D(64, k2, padding='same', kernel_regularizer=l2(1e-4))(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Dropout(0.2)(x)
    # BiLSTM — outputs 128 features (64 units × 2 directions)
    x = Bidirectional(
        LSTM(64, return_sequences=True, dropout=0.2, recurrent_dropout=0.1,
             kernel_regularizer=l2(1e-4), recurrent_regularizer=l2(1e-4))
    )(x)
    # 4-head attention residual (x is 128-dim, attn output matches automatically)
    attn = MultiHeadAttention(num_heads=4, key_dim=32, dropout=0.1)(x, x)
    x = Add()([x, attn])
    x = LayerNormalization()(x)
    # Feed-forward block — expand then project back to 128 to match residual
    ff = Dense(256, activation='relu', kernel_regularizer=l2(1e-4))(x)
    ff = Dropout(0.2)(ff)
    ff = Dense(128, kernel_regularizer=l2(1e-4))(ff)  # must match x dim (128)
    x = Add()([x, ff])
    x = LayerNormalization()(x)
    x = GlobalAveragePooling1D()(x)
    x = Dropout(0.4)(x)
    out = Dense(nc_, activation='softmax', kernel_regularizer=l2(1e-4))(x)
    return Model(inp, out)

def _build_eegnet(ts, nc, nc_):
    kl = max(8, min(64, ts))
    F1, D, F2 = 8, 2, 16
    # Pool sizes must not exceed the spatial dimension remaining after each stage.
    # EEGNet paper uses pool1=4, pool2=8.  We clamp to the actual dimension so
    # the model is valid for any timestep length (e.g. ts=128 would crash with ts//8=16).
    pool1 = min(4, ts)
    pool2 = min(8, max(1, ts // pool1))
    inp = Input(shape=(nc, ts, 1))
    x = BatchNormalization()(Conv2D(F1, (1, kl), padding='same', use_bias=False)(inp))
    x = BatchNormalization()(DepthwiseConv2D((nc, 1), depth_multiplier=D, use_bias=False)(x))
    x = Dropout(0.25)(AveragePooling2D((1, pool1))(Activation('elu')(x)))
    x = BatchNormalization()(SeparableConv2D(F2, (1, max(3, kl // 2)), padding='same', use_bias=False)(x))
    x = Dropout(0.25)(AveragePooling2D((1, pool2))(Activation('elu')(x)))
    return Model(inp, Dense(nc_, activation='softmax')(Flatten()(x)))

# ── Model registry ────────────────────────────────────────────────────────────
_DL_MODELS = {
    '1D CNN':               ('seq', _build_cnn1d),
    'LSTM':                 ('seq', _build_lstm),
    'EEGNet':               ('eeg', _build_eegnet),
    'CNN-BiLSTM+Attention': ('seq', _build_cnn_bilstm),
}

# ── LOSO loop ─────────────────────────────────────────────────────────────────
logo_dl        = LeaveOneGroupOut()
n_splits_dl    = logo_dl.get_n_splits(groups=groups_all)
loso_dl_scores = {name: [] for name in _DL_MODELS}
loso_dl_preds  = {name: [] for name in _DL_MODELS}
loso_dl_trues  = []


MAX_FOLDS = 15  # Limit folds to avoid RAM exhaustion
print(f"\nRunning LOSO over {n_splits_dl} folds for {len(_DL_MODELS)} DL models...\n")

for fold, (train_idx, test_idx) in enumerate(logo_dl.split(X_seq_all, y_all, groups=groups_all)):
    if fold >= MAX_FOLDS:
        break
    left_out_pid = int(groups_all[test_idx][0])

    # 1. Raw sequence splits
    Xseq_tr_raw = X_seq_all[train_idx]
    Xseq_te_raw = X_seq_all[test_idx]
    y_tr = y_all[train_idx]
    y_te = y_all[test_idx]
    g_tr = groups_all[train_idx]

    # 2. Normalise per-channel using TRAIN statistics only (no leakage into test)
    seq_mean = Xseq_tr_raw.mean(axis=(0, 1), keepdims=True)
    seq_std  = Xseq_tr_raw.std(axis=(0, 1),  keepdims=True) + 1e-6
    Xseq_tr  = (Xseq_tr_raw - seq_mean) / seq_std
    Xseq_te  = (Xseq_te_raw - seq_mean) / seq_std

    # 3. Validation split: hold out one training participant for early stopping.
    #    Using (fold+1) % n avoids repeating the same participant for fold 0 and fold 30.
    unique_tr_pids = np.unique(g_tr)
    val_pid  = int(unique_tr_pids[(fold + 1) % len(unique_tr_pids)])
    val_mask = (g_tr == val_pid)
    Xseq_val = Xseq_tr[val_mask]
    y_val    = y_tr[val_mask]
    Xseq_tr2 = Xseq_tr[~val_mask]
    y_tr2    = y_tr[~val_mask]

    # 4. Class weights from the reduced training set
    classes_fold = np.arange(_num_classes, dtype=np.int32)
    cw_arr  = compute_class_weight('balanced', classes=classes_fold, y=y_tr2)
    cw_dict = {int(c): float(w) for c, w in zip(classes_fold, cw_arr)}

    # 5. Pre-compute 2-D EEG format once per fold (used by EEGNet / DeepConvNet)
    Xeeg_tr2 = np.transpose(Xseq_tr2, (0, 2, 1))[:, :, :, None]
    Xeeg_val = np.transpose(Xseq_val, (0, 2, 1))[:, :, :, None]
    Xeeg_te  = np.transpose(Xseq_te,  (0, 2, 1))[:, :, :, None]

    for name, (fmt, build_fn) in _DL_MODELS.items():
        Xtr_use  = Xseq_tr2 if fmt == 'seq' else Xeeg_tr2
        Xval_use = Xseq_val if fmt == 'seq' else Xeeg_val
        Xte_use  = Xseq_te  if fmt == 'seq' else Xeeg_te
        # EarlyStopping must be a fresh instance per model — it is stateful
        # (tracks best, wait, best_weights). A shared instance carries stale
        # state from the previous model and will stop training too early.
        es = EarlyStopping(monitor='val_accuracy', patience=10, restore_best_weights=True)
        model = build_fn(_timesteps, _n_channels, _num_classes)
        model.compile(
            optimizer=Adam(3e-4),
            loss=SparseCategoricalCrossentropy(),
            metrics=['accuracy'],
        )
        model.fit(
            Xtr_use, y_tr2,
            validation_data=(Xval_use, y_val),
            epochs=80,
            batch_size=64,
            callbacks=[es],
            class_weight=cw_dict,
            verbose=0,
        )
        _, acc = model.evaluate(Xte_use, y_te, verbose=0)
        loso_dl_scores[name].append(float(acc))
        probs = model.predict(Xte_use, verbose=0)
        fold_preds = np.argmax(probs, axis=1).tolist()
        loso_dl_preds[name].extend(fold_preds)
        # Free memory: del + gc + clear_session releases TF graph state each model.
        # name counter to zero, causing layer-name collisions across folds in TF2.
        del model
        gc.collect()
        tf.keras.backend.clear_session()  # Fix 2: release TF graph memory each fold

    loso_dl_trues.extend(y_te.tolist())
    print(f"  Fold {fold+1}/{n_splits_dl}  (left out participant {left_out_pid})")

# ── Summary ───────────────────────────────────────────────────────────────────
print("\n" + "=" * 55)
print("   LOSO DEEP LEARNING — MEAN ACCURACIES")
print("=" * 55)
for name, scores in loso_dl_scores.items():
    m, s = np.mean(scores), np.std(scores)
    print(f"  {name:<28} {m:.4f}  (±{s:.4f})")
print("=" * 55)
print(f"  Chance baseline (4 classes):     0.2500")
print("=" * 55)
````

````python
# ============================================================
# LOSO Full Metrics: F1, Recall, AUC (Classical + DL)
# ============================================================

def loso_full_report(name, y_true, y_pred, num_classes, id_to_emotion):
    """Print classification report + macro AUC for one LOSO model."""
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    # Align lengths (in case folds were cut short)
    n = min(len(y_true), len(y_pred))
    y_true, y_pred = y_true[:n], y_pred[:n]

    labels = list(range(num_classes))
    tnames = [id_to_emotion.get(l, str(l)) for l in labels]

    print(f"{'='*60}")
    print(f"  {name}")
    print(f"{'='*60}")
    print(classification_report(y_true, y_pred, labels=labels,
                                 target_names=tnames, zero_division=0))

    # Macro AUC via one-hot binarization of predictions
    # (no probability scores available for LOSO aggregated preds,
    #  so we use the hard predictions as a proxy — reports hard-pred AUC)
    try:
        y_true_bin = label_binarize(y_true, classes=labels)
        y_pred_bin = label_binarize(y_pred, classes=labels)
        if y_true_bin.shape[1] > 1:
            macro_auc = roc_auc_score(y_true_bin, y_pred_bin,
                                       average='macro', multi_class='ovr')
            print(f"  Macro AUC (hard predictions): {macro_auc:.4f}")
    except Exception as e:
        print(f"  AUC could not be computed: {e}")
    print()


_num_classes_loso = len(id_to_emotion) if 'id_to_emotion' in globals() else int(np.max(loso_classical_trues) + 1)

# Classical LOSO metrics
if 'loso_classical_preds' in globals() and 'loso_classical_trues' in globals():
    print("\n" + "="*60)
    print("  LOSO METRICS — CLASSICAL MODELS")
    print("="*60 + "\n")
    for name in loso_classical_preds:
        loso_full_report(name, loso_classical_trues, loso_classical_preds[name],
                         _num_classes_loso, id_to_emotion)

# DL LOSO metrics
if 'loso_dl_preds' in globals() and 'loso_dl_trues' in globals():
    print("\n" + "="*60)
    print("  LOSO METRICS — DEEP LEARNING MODELS")
    print("="*60 + "\n")
    for name in loso_dl_preds:
        loso_full_report(name, loso_dl_trues, loso_dl_preds[name],
                         _num_classes_loso, id_to_emotion)
````

# **7. Random Split Cross-Validation (Stratified k-Fold)**

Repeated stratified k-fold cross-validation as an alternative to LOSO. Windows are split randomly (stratified by class) across all participants — this is **not** participant-disjoint, so results are typically optimistic compared to LOSO. Run both to understand the gap between in-distribution and cross-subject performance.

Uses the same `x_band_all` / `X_seq_all` / `y_all` arrays built in Section 3.

## 7.1 Random Split CV — Classical Models

````python
# ==============================================================================
# Random Split Cross-Validation — Classical ML Models
# ==============================================================================
# Uses StratifiedKFold so every fold keeps the same class ratio as the full dataset.
# Scaler is fit on the training fold only (no leakage into the test fold).
# ==============================================================================

from sklearn.model_selection import StratifiedKFold

RS_N_SPLITS    = 10   # Number of folds
RS_RANDOM_STATE = 42  # Reproducibility seed

skf = StratifiedKFold(n_splits=RS_N_SPLITS, shuffle=True, random_state=RS_RANDOM_STATE)

print(f"Starting Random Split (Stratified {RS_N_SPLITS}-Fold) CV for Classical Models...\n")

rs_classical_scores = {name: [] for name in models.keys()}
rs_classical_preds  = {name: [] for name in models.keys()}
rs_classical_trues  = []

for fold, (train_idx, test_idx) in enumerate(skf.split(x_band_all, y_all)):
    x_tr_fold, x_te_fold = x_band_all[train_idx], x_band_all[test_idx]
    y_tr_fold, y_te_fold = y_all[train_idx], y_all[test_idx]

    # Scale using TRAINING fold statistics only (no leakage)
    rs_scaler = RobustScaler()
    x_tr_fold = rs_scaler.fit_transform(x_tr_fold)
    x_te_fold = rs_scaler.transform(x_te_fold)

    for name, orig_model in models.items():
        model_clone = clone(orig_model)
        model_clone.fit(x_tr_fold, y_tr_fold)
        preds = model_clone.predict(x_te_fold)
        acc = accuracy_score(y_te_fold, preds)
        rs_classical_scores[name].append(acc)
        rs_classical_preds[name].extend(preds.tolist())

    rs_classical_trues.extend(y_te_fold.tolist())
    print(f"  Fold {fold+1}/{RS_N_SPLITS} complete.")

print("\n" + "="*55)
print("   RANDOM SPLIT CV — CLASSICAL MEAN ACCURACIES")
print("="*55)
for name, scores in rs_classical_scores.items():
    m_acc = np.mean(scores)
    s_acc = np.std(scores)
    print(f"  {name:<25} {m_acc:.4f}  (±{s_acc:.4f})")
print("="*55)
````

## 7.2 Random Split CV — Deep Learning Models

````python
# ==============================================================================
# Random Split Cross-Validation — Deep Learning Models
# ==============================================================================
# Same StratifiedKFold split applied to the raw EEG sequence data (X_seq_all).
# Per-channel z-normalisation is fit on the training fold only.
# EarlyStopping uses a held-out 10% of the training fold for validation.
# Each model is rebuilt from scratch every fold to avoid weight leakage.
# ==============================================================================

print(f"Starting Random Split (Stratified {RS_N_SPLITS}-Fold) CV for DL Models...")
print(f"Models: {list(_DL_MODELS.keys())}\n")

rs_dl_scores = {name: [] for name in _DL_MODELS}
rs_dl_preds  = {name: [] for name in _DL_MODELS}
rs_dl_trues  = []

skf_dl = StratifiedKFold(n_splits=RS_N_SPLITS, shuffle=True, random_state=RS_RANDOM_STATE)

for fold, (train_idx, test_idx) in enumerate(skf_dl.split(X_seq_all, y_all)):
    Xseq_tr_raw = X_seq_all[train_idx]
    Xseq_te_raw = X_seq_all[test_idx]
    y_tr = y_all[train_idx]
    y_te = y_all[test_idx]

    # Per-channel z-norm using TRAINING fold stats only
    seq_mean = Xseq_tr_raw.mean(axis=(0, 1), keepdims=True)
    seq_std  = Xseq_tr_raw.std(axis=(0, 1),  keepdims=True) + 1e-6
    Xseq_tr  = (Xseq_tr_raw - seq_mean) / seq_std
    Xseq_te  = (Xseq_te_raw - seq_mean) / seq_std

    # Hold out 10% of training fold for early stopping
    val_size  = max(1, int(0.10 * len(y_tr)))
    val_idx   = np.random.default_rng(fold).choice(len(y_tr), size=val_size, replace=False)
    tr_mask   = np.ones(len(y_tr), dtype=bool)
    tr_mask[val_idx] = False
    Xseq_val  = Xseq_tr[~tr_mask]
    y_val     = y_tr[~tr_mask]
    Xseq_tr2  = Xseq_tr[tr_mask]
    y_tr2     = y_tr[tr_mask]

    # Class weights
    classes_fold = np.arange(_num_classes, dtype=np.int32)
    cw_arr  = compute_class_weight('balanced', classes=classes_fold, y=y_tr2)
    cw_dict = {int(c): float(w) for c, w in zip(classes_fold, cw_arr)}

    # EEG 2-D format for EEGNet
    Xeeg_tr2 = np.transpose(Xseq_tr2, (0, 2, 1))[:, :, :, None]
    Xeeg_val = np.transpose(Xseq_val, (0, 2, 1))[:, :, :, None]
    Xeeg_te  = np.transpose(Xseq_te,  (0, 2, 1))[:, :, :, None]

    for name, (fmt, build_fn) in _DL_MODELS.items():
        Xtr_use  = Xseq_tr2 if fmt == 'seq' else Xeeg_tr2
        Xval_use = Xseq_val if fmt == 'seq' else Xeeg_val
        Xte_use  = Xseq_te  if fmt == 'seq' else Xeeg_te

        es = EarlyStopping(monitor='val_accuracy', patience=10, restore_best_weights=True)
        model = build_fn(_timesteps, _n_channels, _num_classes)
        model.compile(
            optimizer=Adam(3e-4),
            loss=SparseCategoricalCrossentropy(),
            metrics=['accuracy'],
        )
        model.fit(
            Xtr_use, y_tr2,
            validation_data=(Xval_use, y_val),
            epochs=80,
            batch_size=64,
            callbacks=[es],
            class_weight=cw_dict,
            verbose=0,
        )
        _, acc = model.evaluate(Xte_use, y_te, verbose=0)
        rs_dl_scores[name].append(float(acc))
        probs = model.predict(Xte_use, verbose=0)
        rs_dl_preds[name].extend(np.argmax(probs, axis=1).tolist())
        del model
        gc.collect()
        tf.keras.backend.clear_session()

    rs_dl_trues.extend(y_te.tolist())
    print(f"  Fold {fold+1}/{RS_N_SPLITS} complete.")

print("\n" + "="*55)
print("   RANDOM SPLIT CV — DL MEAN ACCURACIES")
print("="*55)
for name, scores in rs_dl_scores.items():
    m, s = np.mean(scores), np.std(scores)
    print(f"  {name:<28} {m:.4f}  (±{s:.4f})")
print("="*55)
print(f"  Chance baseline (4 classes):     0.2500")
print("="*55)
````

````python
# ── Build thesis_rs_preds dict (mirrors thesis_loso_preds) ──
THESIS_RS_CLASSICAL = ['KNN', 'Random Forest', 'SVM (Linear)']
THESIS_RS_DL        = ['1D CNN', 'LSTM', 'EEGNet', 'CNN-BiLSTM+Attention']
THESIS_RS_ALL       = THESIS_RS_CLASSICAL + THESIS_RS_DL

thesis_rs_preds = {}

if 'rs_classical_preds' in globals() and 'rs_classical_trues' in globals():
    y_true_rs_classical = np.asarray(rs_classical_trues, dtype=int)
    for name in THESIS_RS_CLASSICAL:
        if name in rs_classical_preds and len(rs_classical_preds[name]) > 0:
            thesis_rs_preds[name] = {
                'y_pred': np.asarray(rs_classical_preds[name], dtype=int),
                'y_true': y_true_rs_classical,
            }

if 'rs_dl_preds' in globals() and 'rs_dl_trues' in globals():
    y_true_rs_dl = np.asarray(rs_dl_trues, dtype=int)
    for name in THESIS_RS_DL:
        if name in rs_dl_preds and len(rs_dl_preds[name]) > 0:
            thesis_rs_preds[name] = {
                'y_pred': np.asarray(rs_dl_preds[name], dtype=int),
                'y_true': y_true_rs_dl,
            }

print('RS predictions collected for:', list(thesis_rs_preds.keys()))

# ── Per-model metrics table ──
rs_rows = []
for name, d in thesis_rs_preds.items():
    yp = d['y_pred']
    yt = d['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]
    rs_rows.append({
        'Model': name,
        'Accuracy (RS CV)': accuracy_score(yt, yp),
        'Precision (macro)': precision_score(yt, yp, average='macro', zero_division=0),
        'Recall (macro)': recall_score(yt, yp, average='macro', zero_division=0),
        'F1 (macro)': f1_score(yt, yp, average='macro', zero_division=0),
        'F1 (weighted)': f1_score(yt, yp, average='weighted', zero_division=0),
    })

rs_metrics_df = pd.DataFrame(rs_rows).set_index('Model')
display(rs_metrics_df.round(4))

# ── Plot metrics ──
metrics_rs_plot = ['Accuracy (RS CV)', 'Precision (macro)', 'Recall (macro)', 'F1 (macro)']
colors_rs_m = plt.cm.Set2(np.linspace(0, 1, len(rs_metrics_df)))

for met in metrics_rs_plot:
    if met not in rs_metrics_df.columns:
        continue
    vals = rs_metrics_df[met].values.astype(float)
    fig, ax = plt.subplots(figsize=(8, 5))
    bars = ax.bar(rs_metrics_df.index, vals, color=colors_rs_m)
    ax.set_title(f'Thesis Models (Random Split CV) — {met}', fontsize=12)
    ax.set_ylabel(met)
    ax.set_ylim(0, 1.05)
    ax.tick_params(axis='x', rotation=35)
    for bar, value in zip(bars, vals):
        if not np.isnan(value):
            ax.text(bar.get_x() + bar.get_width() / 2, value + 0.02,
                    f'{value:.3f}', ha='center', va='bottom', fontsize=8)
    plt.tight_layout()
    plt.show()
````

````python
# ── ROC curves — Random Split (aggregated hard predictions) ──

for name in THESIS_RS_ALL:
    if name not in thesis_rs_preds:
        print(f'Skipping {name}: no RS predictions found.')
        continue

    yp = thesis_rs_preds[name]['y_pred']
    yt = thesis_rs_preds[name]['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]

    y_bin_true = label_binarize(yt, classes=np.arange(n_classes_eval))
    y_bin_pred = label_binarize(yp, classes=np.arange(n_classes_eval))

    fig, ax = plt.subplots(figsize=(7, 6))
    for c in range(n_classes_eval):
        fpr, tpr, _ = roc_curve(y_bin_true[:, c], y_bin_pred[:, c])
        ax.plot(fpr, tpr, lw=1.5, label=f'{class_labels[c]} (AUC={auc(fpr, tpr):.2f})')
    fpr_m, tpr_m, _ = roc_curve(y_bin_true.ravel(), y_bin_pred.ravel())
    ax.plot(fpr_m, tpr_m, 'k--', lw=2, label=f'micro (AUC={auc(fpr_m, tpr_m):.2f})')
    ax.plot([0, 1], [0, 1], 'k:', alpha=0.4)
    ax.set_title(f'ROC Curves (RS CV hard preds) — {name}', fontsize=12)
    ax.set_xlabel('False Positive Rate (FPR)')
    ax.set_ylabel('True Positive Rate (TPR)')
    ax.legend(fontsize=8, loc='lower right')
    ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
    plt.close(fig)

# ── Combined micro-ROC overlay ──
plt.figure(figsize=(10, 8))
for name in THESIS_RS_ALL:
    if name not in thesis_rs_preds:
        continue
    yp = thesis_rs_preds[name]['y_pred']
    yt = thesis_rs_preds[name]['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]
    y_bin_true = label_binarize(yt, classes=np.arange(n_classes_eval))
    y_bin_pred = label_binarize(yp, classes=np.arange(n_classes_eval))
    fpr, tpr, _ = roc_curve(y_bin_true.ravel(), y_bin_pred.ravel())
    plt.plot(fpr, tpr, lw=2, label=f'{name} (micro AUC={auc(fpr, tpr):.3f})')
plt.plot([0, 1], [0, 1], 'k--', label='Chance')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Combined micro-average ROC — Random Split CV (all thesis models)')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# ── Confusion matrices — Random Split aggregated predictions ──

for name in THESIS_RS_ALL:
    if name not in thesis_rs_preds:
        print(f'Skipping {name}: no RS predictions found.')
        continue

    yp = thesis_rs_preds[name]['y_pred']
    yt = thesis_rs_preds[name]['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]

    # 1. Raw Counts Confusion Matrix
    fig1, ax1 = plt.subplots(figsize=(6, 5))
    cm_raw = confusion_matrix(yt, yp, labels=np.arange(n_classes_eval))
    sns.heatmap(cm_raw, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_labels, yticklabels=class_labels, ax=ax1)
    ax1.set_title(f'Confusion Matrix — {name}\n(Random Split CV)', fontsize=11)
    ax1.set_xlabel('Predicted Labels')
    ax1.set_ylabel('True Labels')
    plt.tight_layout()
    plt.show()
    plt.close(fig1)

    # 2. Normalised Confusion Matrix
    fig2, ax2 = plt.subplots(figsize=(6, 5))
    cm_norm = confusion_matrix(yt, yp, labels=np.arange(n_classes_eval), normalize='true')
    sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Oranges',
                xticklabels=class_labels, yticklabels=class_labels,
                ax=ax2, vmin=0, vmax=1)
    ax2.set_title(f'Normalized Confusion Matrix — {name}\n(RS CV — True-Class Recall)', fontsize=11)
    ax2.set_xlabel('Predicted Labels')
    ax2.set_ylabel('True Labels')
    plt.tight_layout()
    plt.show()
    plt.close(fig2)
````

````python
# ── Per-class precision / recall / F1 heatmap — Random Split ──

prf_rs_rows = []
for name in THESIS_RS_ALL:
    if name not in thesis_rs_preds:
        continue
    yp = thesis_rs_preds[name]['y_pred']
    yt = thesis_rs_preds[name]['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]
    p, r, f, _ = precision_recall_fscore_support(
        yt, yp, labels=np.arange(n_classes_eval), zero_division=0
    )
    for i, lbl in enumerate(class_labels):
        prf_rs_rows.append({'Model': name, 'Class': lbl, 'Precision': p[i], 'Recall': r[i], 'F1': f[i]})

prf_rs_df = pd.DataFrame(prf_rs_rows)
for metric in ['Precision', 'Recall', 'F1']:
    pivot = prf_rs_df.pivot(index='Model', columns='Class', values=metric)
    plt.figure(figsize=(8, 5))
    sns.heatmap(pivot, annot=True, fmt='.2f', cmap='YlGnBu', vmin=0, vmax=1)
    plt.title(f'Per-class {metric} — Random Split CV')
    plt.tight_layout()
    plt.show()

# Classification reports
for name in THESIS_RS_ALL:
    if name not in thesis_rs_preds:
        continue
    yp = thesis_rs_preds[name]['y_pred']
    yt = thesis_rs_preds[name]['y_true']
    n = min(len(yt), len(yp))
    yt, yp = yt[:n], yp[:n]
    print('=' * 60)
    print(f'{name}  [Random Split CV]')
    print('=' * 60)
    print(classification_report(yt, yp, target_names=class_labels, zero_division=0))
````

## 7.5 LOSO vs Random Split — Side-by-Side Comparison

The gap between RS CV and LOSO accuracy reveals how much models rely on seeing windows from the same participant during training (subject-specific patterns). A large gap means the model generalises poorly to new subjects.

````python
# ── LOSO vs Random Split accuracy comparison ──

comparison_names = []
loso_vals  = []
rs_vals    = []

all_model_names = list(dict.fromkeys(
    list(getattr(thesis_loso_preds, 'keys', dict().keys)() if isinstance(thesis_loso_preds, dict) else []) +
    list(thesis_rs_preds.keys())
))

# Safely collect both
for name in THESIS_RS_ALL:
    loso_acc = None
    rs_acc   = None
    if name in thesis_loso_preds:
        d = thesis_loso_preds[name]
        n = min(len(d['y_true']), len(d['y_pred']))
        loso_acc = accuracy_score(d['y_true'][:n], d['y_pred'][:n])
    if name in thesis_rs_preds:
        d = thesis_rs_preds[name]
        n = min(len(d['y_true']), len(d['y_pred']))
        rs_acc = accuracy_score(d['y_true'][:n], d['y_pred'][:n])
    if loso_acc is not None or rs_acc is not None:
        comparison_names.append(name)
        loso_vals.append(loso_acc if loso_acc is not None else np.nan)
        rs_vals.append(rs_acc if rs_acc is not None else np.nan)

if comparison_names:
    x = np.arange(len(comparison_names))
    w = 0.35
    fig, ax = plt.subplots(figsize=(14, 6))
    bars1 = ax.bar(x - w/2, loso_vals, width=w, label='LOSO CV', color='steelblue', alpha=0.85)
    bars2 = ax.bar(x + w/2, rs_vals,   width=w, label='Random Split CV', color='coral', alpha=0.85)
    ax.set_xticks(x)
    ax.set_xticklabels(comparison_names, rotation=30, ha='right')
    ax.set_ylim(0, 1.10)
    ax.set_ylabel('Accuracy')
    ax.set_title('LOSO vs Random Split CV — Accuracy per Model')
    ax.legend()
    ax.axhline(0.25, color='gray', ls=':', alpha=0.6, label='Chance (4-class)')
    for bar, val in zip(bars1, loso_vals):
        if not np.isnan(val):
            ax.text(bar.get_x() + bar.get_width()/2, val + 0.01, f'{val:.3f}',
                    ha='center', va='bottom', fontsize=7)
    for bar, val in zip(bars2, rs_vals):
        if not np.isnan(val):
            ax.text(bar.get_x() + bar.get_width()/2, val + 0.01, f'{val:.3f}',
                    ha='center', va='bottom', fontsize=7)
    plt.tight_layout()
    plt.show()

    # Gap (RS - LOSO) bar chart
    gaps_comp = [r - l if not np.isnan(r) and not np.isnan(l) else np.nan
                 for r, l in zip(rs_vals, loso_vals)]
    gap_colors = ['crimson' if g > 0.15 else 'goldenrod' if g > 0.08 else 'seagreen'
                  for g in gaps_comp if not np.isnan(g)]

    fig2, ax2 = plt.subplots(figsize=(12, 4))
    valid = [(n, g) for n, g in zip(comparison_names, gaps_comp) if not np.isnan(g)]
    ax2.bar([v[0] for v in valid], [v[1] for v in valid], color=gap_colors)
    ax2.axhline(0.10, color='red', ls='--', alpha=0.6, label='Gap threshold 0.10')
    ax2.axhline(0, color='black', lw=0.8)
    ax2.set_ylabel('RS CV − LOSO Accuracy')
    ax2.set_title('Generalisation gap: RS CV minus LOSO accuracy (larger = more subject-specific)')
    ax2.tick_params(axis='x', rotation=30)
    ax2.legend()
    plt.tight_layout()
    plt.show()

    # Summary table
    comp_df = pd.DataFrame({
        'Model': comparison_names,
        'LOSO Acc': loso_vals,
        'RS CV Acc': rs_vals,
        'Gap (RS-LOSO)': gaps_comp,
    }).set_index('Model')
    display(comp_df.round(4))
else:
    print('Run LOSO (Section 5/6/8) and RS CV (Section 7) cells first.')
````
